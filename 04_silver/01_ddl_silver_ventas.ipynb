```python
%sql
create schema if not exists fabriperezchauca.silver
comment 'Datos limpios, tipados y validados';

create or replace table fabriperezchauca.silver.ventas (
    fecha            date           comment 'Fecha de la venta. Año corregido a 2026 cuando estaba mal digitado',
    categoria        string         comment 'Valores: accesorios, uniformes, equipo tactico, equipamiento campo, calzado, otros',
    producto         string         comment 'Nombre del producto en minúsculas y sin espacios extra',
    talla            string         comment 'Sin cambios. Opcional: ~98 % null',
    cantidad         INT            comment 'Unidades vendidas',
    precio_original  decimal(10,2)  comment 'Precio de lista por unidad. Sin símbolo $ ni espacios',
    precio_venta     decimal(10,2)  comment 'Precio cobrado por unidad (de precio_dado). Sin símbolo $ ni espacios',
    descuento        decimal(10,2)  comment 'cantidad x (precio_original - precio_venta) cuando es positivo, si no 0. Recalculado',
    recargo          decimal(10,2)  comment 'cantidad x (precio_venta - precio_original) cuando es positivo, si no 0. Recalculado',
    total            decimal(10,2)  comment 'cantidad x precio_venta. Recalculado, no se toma de la fuente',
    forma_pago       string         comment 'Valores: de una, efectivo, transferencia, tarjeta de credito',
    observaciones    string         comment 'Sin cambios. Opcional: ~98 % null',

    _source_table         string    default 'fabriperezchauca.bronze.ventas' comment 'Tabla origen',
    _processing_timestamp timestamp default CURRENT_TIMESTAMP()              comment 'Timestamp de procesamiento'
)
using delta
tblproperties('delta.feature.allowColumnDefaults' = 'supported')
comment 'Ventas limpias. Grano: una fila = un producto vendido (con su cantidad) en una fecha. No representa una venta completa'
```
