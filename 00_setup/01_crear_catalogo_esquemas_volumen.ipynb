```python
%sql
-- Catálogo del proyecto
CREATE CATALOG IF NOT EXISTS fabriperezchauca
COMMENT 'Datos de ventas del negocio Fabriperez Chauca (uniformes y accesorios policiales)';

USE CATALOG fabriperezchauca;

-- Esquemas
CREATE SCHEMA IF NOT EXISTS landing
COMMENT 'Archivos originales tal como los entrega el negocio, sin modificar';

CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Datos del archivo en tablas, todas las columnas como string, con metadatos de ingesta';

-- Volumen para los archivos crudos
CREATE VOLUME IF NOT EXISTS landing.archivos_crudos
COMMENT 'Excel original y su exportación a CSV UTF-8';
```
